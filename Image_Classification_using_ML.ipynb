```python
# core parts
# pixel values = 0-255

input = [] # To collect input
output = [] # To collect output
categories = ['Dog','Cat'] # should be in the same order as listed
datadir = '/content/drive/MyDrive/Data'
```

```python
for i in categories: # i = Dog
  print("Loading Categories..")
  path=os.path.join(datadir,i) # /content/drive/MyDrive/Data/,Dog
  for img in os.listdir(path):
    img_read=imread(os.path.join(path,img))
    img_res = resize(img_read,(150,150,3))
    input.append(img_res.flatten())
    output.append(categories.index(i))
  print('Loaded Completely')

```

```python
input=np.array(input)
output= np.array(output)
```

```python
df = pd.DataFrame(input)
df
```

```python
df['target']=output
df
```

```python
x = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test =train_test_split(x,y,test_size=0.20,random_state=42)
x_train
```

```python
#  Model Creation
# SVM -(support vector Machine)
# SVM algorithm more used for image classification

from sklearn.svm import SVC
model = SVC()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)# performance in 50-60 range
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score
score = accuracy_score(y_test,y_pred)
score
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_pred)
cm
```
